### 1. Describe the concept of Workflows, Processes and Channels we deal with in Nextflow

Source seqera Docs


Workflows:
In Nextflow, a workflow is a specialized function for composing processes and dataflow logic:

An entry workflow is the entrypoint of a pipeline. It can take parameters as inputs using the params block, and it can publish outputs using the output block.

A named workflow is a workflow that can be called by other workflows. It can define its own inputs and outputs, which are called takes and emits.

Both entry workflows and named workflows can contain dataflow logic such as calling processes, workflows, and channel operators.




Processes:
In Nextflow, a process is a specialized function for executing scripts in a scalable and portable manner.





Channels:
A channel carries data (values or files) from one process to the next. The items go through the channel one after another, and a process takes each item and works on it. There are two types: a queue channel holds many items, and each item is used only once. A value channel holds one value that can be used many times.

The groupTuple operator collects lists (i.e. tuples) from a source channel into groups based on a grouping key. A new tuple is emitted for each distinct key -> it returns a queue channel

```python
# Task 12 - Create a channel that joins the input to the output channel. What do you notice?
!nextflow run basic_channel_operations.nf --step 12
```

The join operator emits the inner product of two source channels using a matching key and returns a channel.

Task 12 - What do you notice compared to Task 11?

In Task 11, groupTuple collects all values with the same key from one channel into a list, for example [1, [V, f, B]]. In Task 12, join combines two channels by the same key, but it takes only one item from each channel per key, for example [1, V, f]. Items without a matching partner, like [1, B], [3, 33] and [2, E], are dropped. So join gives fewer results, and the values are not grouped into a list
