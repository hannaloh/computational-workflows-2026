### What is a container?

Containers are an abstraction that packages code and dependencies together. Many containers can run on the same computer at the same time. Each one runs separately, but they all share the computer's operating system. This is why containers are much smaller and faster than virtual machines, which each need a full operating system of their own.

### Why do we use containers?

Containers are used because they package a tool together with everything it needs: the version, libraries and settings. So the tool works the same on every computer, and everyone gets the same results -> reproducible. One don't have to install anything manually, and there are no version conflicts.

### What is a docker image?

Source: https://docs.docker.com/get-started/docker-concepts/the-basics/what-is-an-image/#explanation

Seeing as a container is an isolated process, where does it get its files and configuration? How do you share those environments?

That's where container images come in. A container image is a standardized package that includes all of the files, binaries, libraries, and configurations to run a container.

There are two important principles of images:

1. Images are immutable. Once an image is created, it can't be modified. You can only make a new image or add changes on top of it.

2. Container images are composed of layers. Each layer represents a set of file system changes that add, remove, or modify files.

These two principles let you to extend or add to existing images

### Compare the file to last weeks fastqc results, are they identical?


```python
import zipfile, glob

def fastqc_data(zip_path):
    with zipfile.ZipFile(zip_path) as z:
        name = [n for n in z.namelist() if n.endswith("fastqc_data.txt")][0]
        return z.read(name).decode().splitlines()

old = glob.glob("../day_02/results/fastqc/raw/*SRX19144486*_1_fastqc.zip")
new = glob.glob("fastqc_results/*_fastqc.zip")
print("old:", old)
print("new:", new)

old_lines, new_lines = fastqc_data(old[0]), fastqc_data(new[0])
print("Version old:", old_lines[0])
print("Version new:", new_lines[0])

# compare all lines except the file name
diff = [(a, b) for a, b in zip(old_lines, new_lines) if a != b and not a.startswith("Filename")]
print("Different lines:", len(diff))
for a, b in diff[:10]:
    print("old:", a, "\nnew:", b)
```

```python
import zipfile, glob
import pandas as pd

files = {
    "manual": glob.glob("fastqc_manual/*_fastqc.zip")[0],
    "docker": glob.glob("fastqc_results/*_fastqc.zip")[0],
    "pipeline": "../day_02/results/fastqc/raw/SRX19144486_raw_1_fastqc.zip",
}

def read_from_zip(path, name):
    with zipfile.ZipFile(path) as z:
        f = [n for n in z.namelist() if n.endswith(name)][0]
        return z.read(f).decode().splitlines()

# 1) Quality summary (PASS / WARN / FAIL per module)
summary = {}
for label, path in files.items():
    lines = read_from_zip(path, "summary.txt")
    summary[label] = {l.split("\t")[1]: l.split("\t")[0] for l in lines}
summary_df = pd.DataFrame(summary)
display(summary_df)

# 2) Basic statistics
stats = {}
for label, path in files.items():
    lines = read_from_zip(path, "fastqc_data.txt")
    start = lines.index(">>Basic Statistics\tpass") if ">>Basic Statistics\tpass" in lines else [i for i, l in enumerate(lines) if l.startswith(">>Basic Statistics")][0]
    block = []
    for l in lines[start + 2:]:
        if l.startswith(">>END_MODULE"):
            break
        k, v = l.split("\t")
        block.append((k, v))
    stats[label] = dict(block)
display(pd.DataFrame(stats))
```

All three FastQC runs (manual, Docker and pipeline) give the same quality results. The read quality is good, because "Per base sequence quality" is PASS. Some modules show WARN or FAIL, which is normal for RNA-seq data. The basic statistics (number of reads, read length, GC content) are also the same

### Is the fastqc version identical?

Yes. Both runs used FastQC version 0.12.1, and the FastQC data files (fastqc_data.txt inside the zip files) are identical except for the file name: all quality metrics are similiar.

This shows that the same tool version on the same input gives the same result, no matter whether it is run inside a container or by nf-core/rnaseq.

### Explain the RUN and ENV lines you added to the file

RUN line (RUN apt-get update && apt-get install -y curl cowsay && rm -rf /var/lib/apt/lists/*)

- RUN executes commands only once, while the image is being built.
- apt-get update downloads the newest list of available Debian packages.
- apt-get install -y curl cowsay installs the programs curl and cowsay. The -y answers "yes" automatically, because nobody can type an answer during the build.
- rm -rf /var/lib/apt/lists/* deletes the package lists afterwards, so the image stays small.
- && connects the commands. They run one after another and are saved as one layer of the image.

ENV line (ENV PATH="$PATH:/usr/games")

- ENV sets an environment variable inside the container.
- On Debian, cowsay is installed in the folder /usr/games. This folder is not in the PATH by default.
- PATH is the list of folders where the system looks for programs.
- Adding /usr/games to PATH makes it possible to run cowsay directly. Without this line, the command would not be found.

No. Most bioinformatics tools already have ready-made images, for example on BioContainers (quay.io/biocontainers) or Seqera Containers. You just pull the image with a specific version tag and run it. You only build your own image if no image exists or you need a special setup.

### What is https://biocontainers.pro ?


BioContainers is a community-driven project that builds, hosts, and maintains container images for scientific software. We maintain more than 100,000 containers for over 12,000 tools. Our goal is to make software easy to discover, install, and run reproducible across laptops, clusters, clouds, and workflow systems.

## Are there other ways to create Docker (or Apptainer) images?

What is https://seqera.io/containers/ ?

Yes. Besides writing your own Dockerfile and pulling ready-made images from BioContainers, you can use Seqera Containers.

Seqera Containers is a free web service from Seqera, the company behind Nextflow. You choose one or more tools from Bioconda, conda-forge or PyPI, and it builds a container image for you automatically, either as a Docker image or as a Singularity/Apptainer image. You don't need to write a Dockerfile.
